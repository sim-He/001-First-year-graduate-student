# 第三题：手动实现K-means

实验内容：
1. 实现一个K-means聚类算法，并在Breast_Cancer_Wisconsin数据集上完成聚类任务
2. 计算外部指标FMI和NMI
3. 对聚类结果可视化

我们要实现一个K-means算法，也称为原型聚类算法。

## 初始化

K-means在实现的时候，首先需要选取类簇中心。类簇中心的选取方法有很多，我们这里使用最简单的方法，随机选取。也就是，从给定的待聚类的样本中，随机选取 $K$ 个样本，作为 $K$ 个类簇的中心。

## 优化

选取类中心后，就需要不断的调成类中心的位置，开始优化过程，优化主要分为两步：

### 第一步

计算所有样本到 $K$ 个类中心的距离。每个样本，选择距自己最近的类中心作为自己属于的类簇。（这里的距离我们选择欧式距离）

### 第二步

针对第一步分出来的 $K$ 个类簇，计算每个类簇内样本的均值，将计算得到的 $K$ 个均值向量，作为这 $K$ 个类簇新的中心。

### 然后循环第一步和第二步，直至一定的迭代次数，或类中心无显著的位置改变为止。

```python
# 同时支持从项目根目录和 solution 目录启动 Notebook。
from pathlib import Path

data_path = Path('data/Breast_Cancer_Wisconsin/data')
if not data_path.is_file():
    data_path = Path('../data/Breast_Cancer_Wisconsin/data')
data = pd.read_csv(data_path)
# B 表示良性，M 表示恶性；真实标签只用于评价，不参与训练。
data['diagnosis'] = data['diagnosis'].map({'B': 1, 'M': -1})
```

# 3. 欧式距离的实现

给定向量 $x \in \mathbb{R}^m$，$y \in \mathbb{R}^m$，两个向量的欧式距离定义为：

$$
E(x, y) = \sqrt{\sum^m_{i = 1} (x_i - y_i)^2}
$$

其中, $i$ 表示向量的第 $i$ 个分量。

我们要实现一个可以计算多个样本组成的矩阵 $X$，与某一个类中心 $y$ 之间欧氏距离的函数。

给定输入矩阵 $X \in \mathbb{R}^{n \times m}$，其中 $n$ 是样本数，$m$ 是特征数，给定输入的类簇中心 $y \in \mathbb{R}^m$。

我们要计算 $n$ 个样本到某一类簇中心 $y$ 的欧式距离，最后的结果是 $E \in \mathbb{R}^{n}$，每个元素表示矩阵 $X$ 中的每个样本到类中心 $y$ 的欧式距离。

```python
def compute_distance(X, y):
    """计算 X 中每行样本与向量 y 的欧氏距离，返回形状为 (n,) 的数组。"""
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float)
    # 按特征维度累加平方差，再开平方。
    distance = np.sqrt(np.sum((X - y) ** 2, axis=1))
    return distance
```

```python
# 测试样例
print(compute_distance(np.array([[0, 0], [0, 1]]), np.array([0, 1]))) # [ 1.  0.]
print(compute_distance(np.array([[0, 0], [0, 1]]), np.array([1, 1]))) # [ 1.41421356  1.        ]
```

下面开始实现K-means聚类算法

```python
class myKmeans:
    def __init__(self, n_clusters, max_iter=100, random_state=42, tol=1e-6):
        """设置簇数、最大迭代次数、随机种子和中心移动的收敛阈值。"""
        self.n_clusters = n_clusters
        self.max_iter = int(max_iter)
        self.random_state = random_state
        self.tol = tol
        self.centroids = None

    def choose_centroid(self, X):
        """随机选择 K 个不重复的样本作为初始中心，返回 (K, m) 数组。"""
        rng = np.random.default_rng(self.random_state)
        indices = rng.choice(len(X), self.n_clusters, replace=False)
        return X[indices, :].copy()

    def compute_label(self, X):
        """计算每个样本到所有中心的距离，返回最近中心的编号，形状为 (n,)。"""
        if self.centroids is None:
            raise ValueError('请先调用 fit 训练模型。')
        X = np.asarray(X, dtype=float)
        distances = np.empty((len(X), self.n_clusters))
        for index in range(len(self.centroids)):
            # 第 index 列保存每个样本到第 index 个中心的欧氏距离。
            distances[:, index] = compute_distance(X, self.centroids[index])
        # 每一行最小距离所在的列即为簇编号。
        labels = np.argmin(distances, axis=1)
        return labels

    def fit(self, X):
        """交替执行最近中心分配和簇内均值更新，直至收敛或达到迭代上限。"""
        X = np.asarray(X, dtype=float)
        if X.ndim != 2 or X.shape[0] == 0 or X.shape[1] == 0 or not np.isfinite(X).all():
            raise ValueError('X 必须是非空且有限的二维数值数组。')
        if not isinstance(self.n_clusters, (int, np.integer)) or not 1 <= self.n_clusters <= len(X):
            raise ValueError('n_clusters 必须是 1 到样本数之间的整数。')
        if self.max_iter < 1 or self.tol < 0:
            raise ValueError('max_iter 必须为正数，tol 不能为负数。')
        self.centroids = self.choose_centroid(X)
        for epoch in range(self.max_iter):
            labels = self.compute_label(X)
            old_centroids = self.centroids.copy()
            for index in range(self.n_clusters):
                members = X[labels == index]
                # 非空簇更新为成员均值；空簇保留原中心，避免空数组均值产生 NaN。
                if len(members) > 0:
                    self.centroids[index, :] = np.mean(members, axis=0)
            self.n_iter_ = epoch + 1
            shift = np.linalg.norm(self.centroids - old_centroids, axis=1)
            if np.max(shift) <= self.tol:
                break
        # 使用最终中心重新分配标签，保证标签和中心一致。
        self.labels_ = self.compute_label(X)
        self.inertia_ = float(np.sum((X - self.centroids[self.labels_]) ** 2))
        return self
```

模型训练

```python
# 初始化一个具有 2 个簇的模型；训练过程仅使用特征。
model = myKmeans(n_clusters=2, max_iter=100, random_state=42)
model.fit(data_x)
# 根据最终中心获取每个样本的簇编号。
y_hat = model.compute_label(data_x)
print(f'迭代次数：{model.n_iter_}，簇内平方和：{model.inertia_:.4f}')
```

聚类结果统计

```python
def getResult(data_y,y_hat):
    true_labels = data_y.reshape(-1)
    cluster = {}
    # 构造簇
    for i in range(len(y_hat)):
        cluster_label = y_hat[i]
        if cluster_label not in cluster:
            cluster[cluster_label] = {}
    # 构造簇内类别标签
    for cluster_label in cluster:
        for true_label in list(set(true_labels)):
            cluster[cluster_label][true_label] = 0
    # 添加簇内数据
    for i in range(len(y_hat)):
            cluster_label = y_hat[i]
            cluster[cluster_label][true_labels[i]] +=1
    # 按照簇序号排序
    cluster = dict(sorted(cluster.items(),key = lambda x:x[0]))
    return cluster
cluster = getResult(data_y,y_hat)
print(cluster)
```

# 5. 聚类结果可视化

```python
def draw_bar(cluster_data):
    true_labels = sorted({label for counts in cluster_data.values() for label in counts})
    cluster_labels = sorted(cluster_data)
    x = np.arange(len(cluster_labels))
    bar_width = 0.35
    plt.figure(figsize=(7, 4))
    for index, label in enumerate(true_labels):
        counts = [cluster_data[c].get(label, 0) for c in cluster_labels]
        offset = (index - (len(true_labels) - 1) / 2) * bar_width
        name = 'Benign (B)' if label == 1 else 'Malignant (M)'
        plt.bar(x + offset, counts, width=bar_width, label=name)
    plt.xticks(x, [f'cluster {c}' for c in cluster_labels])
    plt.xlabel('Cluster')
    plt.ylabel('Number of samples')
    plt.title('Cluster result')
    plt.legend()
    plt.tight_layout()
    plt.show()

draw_bar(cluster)
```

# 6. 评价指标

这里，我们选用两个外部指标，FMI和NMI。

```python
from sklearn.metrics import normalized_mutual_info_score
from sklearn.metrics import fowlkes_mallows_score
```

```python
# FMI 和 NMI 都不要求簇编号与真实标签的数值相同，越接近 1 越好。
fmi = fowlkes_mallows_score(data_y, y_hat)
nmi = normalized_mutual_info_score(data_y, y_hat)
print(f'FMI = {fmi:.6f}')
print(f'NMI = {nmi:.6f}')

# 用前两个原始特征展示聚类分配及最终中心。
plt.figure(figsize=(7, 5))
plt.scatter(data_x[:, 0], data_x[:, 1], c=y_hat, cmap='viridis', s=20, alpha=0.7)
plt.scatter(model.centroids[:, 0], model.centroids[:, 1], c='red', marker='X', s=180, label='Centroids')
plt.xlabel('radius_mean')
plt.ylabel('texture_mean')
plt.title('Manual K-means: projection of 30-feature clustering')
plt.legend()
plt.tight_layout()
plt.show()
```

本次运行结果：7 次迭代后收敛，FMI = 0.791517、NMI = 0.464793，簇内平方和为 77943099.8783。在本次数据和参数设置下，手动实现与 sklearn KMeans 的两个外部指标相同；这不意味着不同数据或初始化时一定得到相同结果。

实验分析：本实现按“随机选取初始中心—分配最近簇—计算簇内均值”进行迭代，未调用 sklearn 的 KMeans。真实诊断标签仅用于统计和评价。散点图只展示 30 维聚类结果在两个特征上的投影，部分重叠不能单独证明高维聚类无效。随机初始化可能导致局部最优；本实验保留原始特征尺度，面积等数值范围较大的特征对欧氏距离贡献更大，可以进一步通过标准化和多次初始化改进。
