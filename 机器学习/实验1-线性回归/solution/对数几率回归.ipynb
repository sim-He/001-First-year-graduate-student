# 对数几率回归

## 实验内容
1. 使用对数几率回归完成Breast_Cancer_Wisconsin结果预测问题。
2. 计算十折交叉验证下的精度(accuracy)，查准率(precision)，查全率(recall)，F1值。

## 评测指标  
1. 精度
2. 查准率
3. 查全率
4. F1

# 1. 读取数据

```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('breast-cancer.csv')
```

# 2. 导入模型

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.model_selection import cross_val_predict
from sklearn import preprocessing
```

```python
data.head()
```

数据处理

```python
data = data.values 
data_x = data[:,2:-1]
data_y = data[:,1:2]
```

```python
data_y = np.reshape(data_y,(-1))
data_y.shape
```

# 4. 训练并预测

请使用预处理后的数据进行模型的训练与预测

```python
# 创建逻辑回归模型
model = LogisticRegression(max_iter = 1000)

# 标准化特征，帮助逻辑回归更快收敛
x_scaled = preprocessing.scale(data_x)

# 十折交叉验证预测
prediction = cross_val_predict(model, x_scaled, data_y, cv = 10)
```

# 5. 评价指标的计算

请你计算模型的四项指标

```python
accuracy = accuracy_score(data_y, prediction)
precision = precision_score(data_y, prediction, pos_label = 'M')
recall = recall_score(data_y, prediction, pos_label = 'M')
f1 = f1_score(data_y, prediction, pos_label = 'M')

print('accuracy:', accuracy)
print('precision:', precision)
print('recall:', recall)
print('f1:', f1)
```

###### 双击此处填写

# 6. 选做：尝试对特征进行变换、筛选、组合后，训练模型并计算十折交叉验证后的四项指标

```python
def evaluate_model(x, model):
    prediction = cross_val_predict(model, x, data_y, cv = 10)
    acc = accuracy_score(data_y, prediction)
    prec = precision_score(data_y, prediction, pos_label = 'M')
    rec = recall_score(data_y, prediction, pos_label = 'M')
    f1 = f1_score(data_y, prediction, pos_label = 'M')
    print('accuracy:', acc)
    print('precision:', prec)
    print('recall:', rec)
    print('f1:', f1)
    print()

# 模型1：原始特征，不进行标准化
model1 = LogisticRegression(solver = 'liblinear', max_iter = 1000)
print('模型1：原始特征')
evaluate_model(data_x, model1)

# 模型2：全部特征标准化
model2 = LogisticRegression(max_iter = 1000)
x_scaled_all = preprocessing.scale(data_x)
print('模型2：标准化全特征')
evaluate_model(x_scaled_all, model2)

# 模型3：标准化后只使用前10个均值类特征
model3 = LogisticRegression(max_iter = 1000)
x_scaled_mean = preprocessing.scale(data_x[:, :10])
print('模型3：标准化 + 均值类特征')
evaluate_model(x_scaled_mean, model3)
```

###### 双击此处填写
1. 模型1的处理流程：原始特征，未标准化，使用 liblinear 求解器
2. 模型2的处理流程：全部特征做 z-score 标准化
3. 模型3的处理流程：标准化后仅使用前 10 个均值类特征
