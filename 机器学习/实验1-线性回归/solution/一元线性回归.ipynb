# 一元线性回归

基本形式
$$\tag{1}
f(\mathbf{x}) = w_1 x_1 + w_2 x_2 + ... + w_d x_d + b
$$
向量形式
$$\tag{2}
f(\mathbf{x}) = \mathbf{w}^\mathrm{T}\mathbf{x}+b
$$
其中$\mathbf{w} = (w_1; w_2; ...; w_d)$.

针对一元线性回归，我们有
$$\tag{3}
f(x) = w x + b$$

## 4. 编写模型

我们以类(class)的形式编写这个模型，python中的类很简单，只不过这个类里面需要调用两个函数，一个是get_w，这个是计算模型w的函数，另一个是get_b，计算模型b的函数，需要大家来完成。

本实验要求使用最小二乘法求解一元线性回归模型  
求解$w$和$b$使均方误差$E_{(w,b)} = \sum^m_{i=1}(y_i - wx_i - b)^2$最小化的过程，称为线性回归模型的最小二乘“参数估计”(parameter estimation)。我们可将$E_{(w,b)}$分别对$w$和$b$求导，得到
$$\tag{4}
\frac{\partial E_{(w,b)}}{\partial w} = 2(w \sum^m_{i=1} x^2_i - \sum^m_{i=1} (y_i - b) x_i),
$$

$$\tag{5}
\frac{\partial E_{(w,b)}}{\partial b} = 2(mb - \sum^m_{i=1}(y_i - w x_i))
$$

然后令式(4)和式(5)为0，可得到$w$和$b$的闭式解(closed-form solution)
$$\tag{6}
w = \frac{\sum^m_{i=1} y_i(x_i - \bar{x})}{\sum^m_{i=1}x^2_i - \frac{1}{m}(\sum^m_{i=1}x_i)^2}
$$

$$\tag{7}
b = \frac{1}{m}\sum^m_{i=1}(y_i - w x_i)
$$
其中，$\bar{x} = \frac{1}{m}\sum^m_{i=1}x_i$为$x$的均值

首先编写求解w的函数，传入的参数就是x和y，都是np.ndarray类型的，或是pd.Series类型的（其实都一样）。我们需要大家在下面完成式(6)和式(7)的求解过程，将计算得到的w和b的值返回

```python
def get_w(x, y):
    '''
    这个函数是计算模型w的值的函数，
    传入的参数分别是x和y，表示数据与标记
    
    Parameter
    ----------
        x: np.ndarray，pd.Series，传入的特征数据

        y: np.ndarray, pd.Series，对应的标记
    
    Returns
    ----------
        w: float, 模型w的值
    '''
    
    # m表示样本的数量
    m = len(x)
    
    # 求x的均值
    x_mean = np.mean(x)
    
    # 求w的分子部分
    numerator = np.sum(y * (x - x_mean))
    
    # 求w的分母部分
    denominator = np.sum(x ** 2) - np.sum(x) ** 2 / m
    
    # 求w
    w = numerator / denominator
    
    # 返回w
    return w
```

```python
def get_b(x, y, w):
    '''
    这个函数是计算模型b的值的函数，
    传入的参数分别是x, y, w，表示数据，标记以及模型的w值
    
    Parameter
    ----------
        x: np.ndarray，pd.Series，传入的特征数据

        y: np.ndarray, pd.Series，对应的标记
        
        w: np.ndarray, pd.Series，模型w的值
    
    Returns
    ----------
        b: float, 模型b的值
    '''
    # 样本个数
    m = len(x)
    
    # 求b
    b = np.sum(y - w * x) / m
    
    # 返回b
    return b
```

下面这个类，就是一个最简单的一元线性回归的类，我们已经帮你实现好了三个方法

```python
class myLinearRegression:
    def __init__(self):
        '''
        类的初始化方法，不需要初始化的参数
        这里设置了两个成员变量，用来存储模型w和b的值
        '''
        self.w = None
        self.b = None
    
    def fit(self, x, y):
        '''
        这里需要编写训练的函数，也就是调用模型的fit方法，传入特征x的数据和标记y的数据
        这个方法就可以求解出w和b
        '''
        self.w = get_w(x, y)
        self.b = get_b(x, y, self.w)
        
    def predict(self, x):
        '''
        这是预测的函数，传入特征的数据，返回模型预测的结果
        '''
        if self.w == None or self.b == None:
            print("模型还未训练，请先调用fit方法训练")
            return 
        
        return self.w * x + self.b
```

## 5. 预测

```python
# 创建一个模型的实例
model1 = myLinearRegression()

# 使用训练集对模型进行训练，传入训练集的TV和标记Sales
model1.fit(train_data['TV'], train_data['Sales'])

# 对测试集进行预测，并将结果存储在变量prediction中
prediction1 = model1.predict(test_data['TV'])
```

## 6. 性能度量

模型训练完成后，还需要在测试集上验证其预测能力，这就需要计算模型的一些性能指标，如MAE和RMSE等。

$$\tag{8}
MAE(\hat{y}, y) = \frac{1}{m} \sum^m_{i=1} \vert \hat{y} - y \vert
$$

$$\tag{9}
RMSE(\hat{y}, y) = \sqrt{\frac{1}{m} \sum^m_{i=1} (\hat{y} - y)^2}
$$
其中，$\hat{y}$是模型的预测值，$y$是真值，$m$是样本数

```python
def MAE(y_hat, y):
    # 计算平均绝对误差：MAE = (1/m) * sum(|y_hat - y|)
    m = len(y)
    return np.sum(np.abs(y_hat - y)) / m
```

```python
def RMSE(y_hat, y):
    # 计算均方根误差：RMSE = sqrt((1/m) * sum((y_hat - y)^2))
    m = len(y)
    return np.sqrt(np.sum((y_hat - y) ** 2) / m)
```

在此计算出模型在测试集上的MAE与RMSE值

```python
mae1 = MAE(prediction1, test_data['Sales'])
rmse1 = RMSE(prediction1, test_data['Sales'])
print("模型1，特征：TV")
print("MAE:", mae1)
print("RMSE:", rmse1)
```

```python
# 创建新的图
plt.figure(figsize = (16, 6))

# 创建子图1
plt.subplot(121) # 121分别代表，生成1行2列个图，这是第1个
# 其中的参数为横轴值，纵轴值，label为此条曲线的标签， '.' 表示画出的图的图形为散点图
plt.plot(train_data['TV'].values, train_data['Sales'].values, '.', label = 'training data')
# '-' 表示画出的图形为折线图
plt.plot(train_data['TV'].values, model1.predict(train_data['TV']), '-', label = 'prediction')
plt.xlabel("TV")
plt.ylabel('Sales')
plt.title("training set")
plt.legend()

# 创建子图2
plt.subplot(122) # 121分别代表，生成1行2列个图，这是第2个
plt.plot(test_data['TV'].values, test_data['Sales'].values, '.', label = 'testing data')
plt.plot(test_data['TV'].values, prediction1, '-', label = 'prediction')
plt.xlabel("TV")
plt.ylabel('Sales')
plt.title("testing set")
plt.legend()
```

通过左右两图的对比，TV 模型的拟合直线在训练集和测试集上都基本贴近样本点，但测试集上的预测误差略大于训练集，说明模型存在一定程度的欠拟合（bias 偏大）：

- **出现的问题**：只用 TV 一个特征解释 Sales，忽略了 Radio、Newspaper 对 Sales 的影响；部分样本点明显偏离拟合直线，这些点可能受其他特征影响较大，或者是离群值，会拉偏最小二乘回归线。
- **改进方案**：
  1. 使用多元线性回归，同时输入 TV、Radio、Newspaper；
  2. 构造多项式特征或特征交叉项，捕捉非线性关系；
  3. 训练前剔除离群值，减少其对最小二乘解的影响；
  4. 对特征进行标准化，并使用交叉验证评估泛化性能。

# 选做：剔除训练集中的离群值(outlier)，然后重新训练模型，观察模型预测性能的变化
###### 提示：可以使用下面的代码处理数据

```python
# 先训练一个初始 TV 模型，用于观察训练集上的残差
model_temp = myLinearRegression()
model_temp.fit(train_data['TV'], train_data['Sales'])

# 计算训练集上的预测残差
residuals = train_data['Sales'] - model_temp.predict(train_data['TV'])

# 将残差绝对值超过 2 倍标准差视为离群值并剔除
threshold = 2 * np.std(residuals)
outlier_mask = residuals.abs() <= threshold
train_data_clean = train_data[outlier_mask]
print('剔除离群值数量:', (~outlier_mask).sum())

# 可视化训练集中的离群值
plt.figure(figsize=(6, 6))
plt.plot(train_data['TV'].values, train_data['Sales'].values, '.', label='training data')
plt.plot(train_data.loc[~outlier_mask, 'TV'].values, train_data.loc[~outlier_mask, 'Sales'].values, 'rx', markersize=10, label='outlier')
plt.xlabel('TV')
plt.ylabel('Sales')
plt.title('training data with outliers')
plt.legend()
```

```python
# 使用剔除离群值后的训练集重新训练模型
model4 = myLinearRegression()
model4.fit(train_data_clean['TV'], train_data_clean['Sales'])

# 在测试集上计算指标
prediction4 = model4.predict(test_data['TV'])
mae4 = MAE(prediction4, test_data['Sales'])
rmse4 = RMSE(prediction4, test_data['Sales'])
print("模型4，特征：TV（剔除离群值后）")
print("MAE:", mae4)
print("RMSE:", rmse4)

# 绘制训练集和测试集预测曲线
plt.figure(figsize=(16, 6))

plt.subplot(121)
plt.plot(train_data_clean['TV'].values, train_data_clean['Sales'].values, '.', label='training data(clean)')
plt.plot(train_data_clean['TV'].values, model4.predict(train_data_clean['TV']), '-', label='prediction')
plt.xlabel('TV')
plt.ylabel('Sales')
plt.title('training set')
plt.legend()

plt.subplot(122)
plt.plot(test_data['TV'].values, test_data['Sales'].values, '.', label='testing data')
plt.plot(test_data['TV'].values, prediction4, '-', label='prediction')
plt.xlabel('TV')
plt.ylabel('Sales')
plt.title('testing set')
plt.legend()
```
