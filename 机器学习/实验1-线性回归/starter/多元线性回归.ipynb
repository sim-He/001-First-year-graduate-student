## 一、多元线性回归
这部分的内容是要求大家完成多元线性回归，我们会先带着大家使用sklearn做一元线性回归的十折交叉验证，多元线性回归大家可以仿照着完成

### 1. 读取数据

```python
import pandas as pd
```

```python
# 读取数据
data = pd.read_csv('advertising.csv')
```

```python
data.head()
```

### 2. 引入模型

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import cross_val_predict
```

### 3. 使用sklearn完成一元线性回归的十折交叉验证验证

#### 创建模型

```python
model = LinearRegression()
```

#### 选取数据

```python
features = ['TV']
x = data[features]
y = data['Sales']
```

#### 做十折交叉验证的预测

```python
prediction = cross_val_predict(model, x, y, cv = 10)
```

这十折交叉验证是按顺序做的，会先将前10%的数据作为测试集，然后会往后顺延到10%到20%，最后将这十份的预测值按顺序拼接后返回

```python
prediction.shape
```

### 4. 计算评价指标

#### MAE

```python
mean_absolute_error(prediction, data['Sales'])
```

#### RMSE

```python
mean_squared_error(prediction, data['Sales']) ** 0.5
```

### 5. 请你选择多种特征进行组合，完成多元线性回归，并对比不同的特征组合，它们训练出的模型在十折交叉验证上MAE与RMSE的差别，至少完成3组

###### 扩展：多项式回归（一元线性回归的扩展），尝试对部分特征进行变换，如将其二次幂，三次幂作为特征输入模型，观察模型在预测能力上的变化
###### 提示：多元线性回归，只要在上方的features这个list中，加入其他特征的名字就可以

```python
# 定义要比较的特征组合
feature_combinations = [
    ['TV', 'Radio', 'Newspaper'],
    ['TV', 'Radio'],
    ['TV', 'Newspaper'],
    ['Radio', 'Newspaper'],
]

for features in feature_combinations:
    x = data[features]
    # 十折交叉验证预测
    prediction = cross_val_predict(model, x, y, cv = 10)
    mae = mean_absolute_error(prediction, y)
    rmse = mean_squared_error(prediction, y) ** 0.5
    print(features)
    print('MAE:', mae)
    print('RMSE:', rmse)
    print()
```

###### 双击此处填写

2. 模型2使用的特征：TV、Radio

4. 模型4使用的特征：Radio、Newspaper

```python
# 选做：多项式回归
data_poly = data.copy()
data_poly['TV2'] = data_poly['TV'] ** 2
data_poly['TV3'] = data_poly['TV'] ** 3

# 对比：仅TV、TV+TV^2、TV+TV^2+TV^3
poly_combinations = [
    ['TV'],
    ['TV', 'TV2'],
    ['TV', 'TV2', 'TV3'],
]

for features in poly_combinations:
    x = data_poly[features]
    prediction = cross_val_predict(model, x, y, cv = 10)
    mae = mean_absolute_error(prediction, y)
    rmse = mean_squared_error(prediction, y) ** 0.5
    print(features)
    print('MAE:', mae)
    print('RMSE:', rmse)
    print()

# 再尝试把 TV、Radio 的二次项加入多元模型
data_poly['Radio2'] = data_poly['Radio'] ** 2
x = data_poly[['TV', 'Radio', 'TV2', 'Radio2']]
prediction = cross_val_predict(model, x, y, cv = 10)
print(['TV', 'Radio', 'TV2', 'Radio2'])
print('MAE:', mean_absolute_error(prediction, y))
print('RMSE:', mean_squared_error(prediction, y) ** 0.5)
```
